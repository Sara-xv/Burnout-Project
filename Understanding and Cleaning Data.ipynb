### Handle Missing values 

```python
missing_cols = [
    'work_life_balance_score',
    'sleep_quality_score',
    'manager_support_score',
    'years_since_last_promotion',
    'monthly_salary_usd',
    'anxiety_score',
    'job_satisfaction_score',
    'social_support_score',
    'stress_level_self_report',
    'depression_score'
]

# 1) Do missing values overlap across columns? (i.e. does one row have multiple columns missing at once?)
miss_matrix = data[missing_cols].isna()
missing_per_row = miss_matrix.sum(axis=1)

print("Rows with 0 missing columns:", (missing_per_row == 0).sum())
print("Rows with exactly 1 missing column:", (missing_per_row == 1).sum())
print("Rows with 2 or more missing columns:", (missing_per_row >= 2).sum())

# 2) Does the missing rate differ across groups? (e.g. role, work_mode, burnout_risk)
group_cols = ['role', 'work_mode', 'burnout_risk']

for group_col in group_cols:
    print(f"\n--- Missing rate (%) per column by {group_col} ---")
    for col in missing_cols:
        rates = data.groupby(group_col)[col].apply(lambda x: x.isna().mean() * 100)
        print(f"\n{col}:")
        print(rates.round(2))
```

```python
# 3) Impute missing values using median imputation

for col in missing_cols:
    data[col] = data[col].fillna(data[col].median())
```

1. Missing Values (real issue — handled)
10 numeric columns have missing values, ranging from 0.77% to 1.68% per column (2,922 rows affected out of 25,000, 0.48% of all cells). The pattern was checked and found to be random (MCAR) — not concentrated in any particular role, work mode, or burnout risk group. This was resolved using median imputation.

### Handle Outlier  and Invalid

2. Invalid Values (false positive — not a real issue)
The report flagged 16,382 values in presenteeism_days_per_month and absenteeism_days_per_month as invalid, using the rule “month must be between 1 and 12.” This rule was incorrectly applied: these columns represent the number of days per month an employee was present/absent (ranging 0–9 and 0–6), not a calendar month value. The _per_month suffix in the column name triggered an incorrect validation rule in the automated tool. No action is needed here — this should simply be documented as a known limitation of the profiling rule, not treated as a data defect.

3. Outliers (mostly expected — not a defect)
16 of 21 numeric columns had statistical outliers flagged by IQR/MAD methods, but their contribution to the quality penalty was minor (1.33/15). Columns like overtime_hours_per_week (13.31% outliers) and absenteeism_days_per_month (6.43% outliers) are highly skewed by nature — some employees genuinely work much more overtime or have much higher absenteeism than others. These aren’t necessarily errors; the IQR method is simply more sensitive to skewed distributions. These can be left as-is or reviewed case-by-case, but they don’t indicate a data integrity problem.
